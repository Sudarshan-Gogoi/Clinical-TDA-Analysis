```python
# =====================================================================
# PERSISTENT HOMOLOGY, PERSISTENCE DIAGRAMS, PERSISTENCE LANDSCAPES
# AND BETTI CURVES USING ALL 42 BIOMARKERS
#
# Updates:
#   * Larger fonts for titles, axes, ticks, legends and annotations.
#   * Larger persistence markers for H0 components and H1 loops.
#   * Separate H0 and H1 landscape figures for every cohort.
#   * High-resolution (600 dpi) PNG output.
# =====================================================================

import os
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import gudhi as gd

warnings.filterwarnings("ignore")


# =====================================================================
# 1. FILE PATHS AND SETTINGS
# =====================================================================

INPUT_FILE = r"D:\Clinical TDA\Data Analysis\cancer clinical data.xlsx"
SHEET_NAME = "Normal and Cancer"
OUTPUT_DIR = r"D:\Clinical TDA\Data Analysis"

if not os.path.exists(INPUT_FILE):
    alternative_file = "/mnt/data/cancer clinical data.xlsx"
    if os.path.exists(alternative_file):
        INPUT_FILE = alternative_file
        OUTPUT_DIR = "/mnt/data"
    else:
        raise FileNotFoundError(
            "Input Excel file not found. Please check INPUT_FILE."
        )

OUTPUT_DIR = os.path.abspath(OUTPUT_DIR)
os.makedirs(OUTPUT_DIR, exist_ok=True)

# All figures are saved directly in this single output directory.
plt.rcParams["savefig.directory"] = OUTPUT_DIR


# =====================================================================
# 2. ANALYSIS PARAMETERS
# =====================================================================

COHORTS = ["Normal", "Bladder", "Ovarian", "Pancreatic"]
EXPECTED_BIOMARKERS = 42
MAX_EDGE_LENGTH = 1.0
MAX_HOMOLOGY_DIMENSION = 2
HOMOLOGY_COEFFICIENT_FIELD = 2
MIN_PERSISTENCE = 0.0

N_RADII = 501
RADII = np.linspace(0, MAX_EDGE_LENGTH, N_RADII)
N_LANDSCAPE_LAYERS = 5


# =====================================================================
# 3. QUALITY PLOTTING SETTINGS
# =====================================================================

plt.rcParams.update({
    "font.family": "Arial",
    "font.size": 17,
    "axes.labelsize": 22,
    "axes.titlesize": 25,
    "legend.fontsize": 17,
    "xtick.labelsize": 18,
    "ytick.labelsize": 18,
    "figure.titlesize": 27,
    "axes.linewidth": 1.7,
    "lines.linewidth": 2.8,
    "savefig.dpi": 600,
    "mathtext.fontset": "dejavusans",
    "axes.unicode_minus": False
})

TITLE_SIZE = 25
AXIS_LABEL_SIZE = 22
TICK_LABEL_SIZE = 18
LEGEND_SIZE = 17
ANNOTATION_SIZE = 15


# =====================================================================
# 4. LOAD DATA
# =====================================================================

print("\n" + "=" * 80)
print("LOADING CLINICAL DATA")
print("=" * 80)

df = pd.read_excel(INPUT_FILE, sheet_name=SHEET_NAME)

print("Input file:", INPUT_FILE)
print("Output directory:", os.path.abspath(OUTPUT_DIR))
print("Total number of patients:", len(df))
print("Number of columns:", len(df.columns))

print("\\nDATASET COLUMN DETAILS")
print("-" * 80)
for column in df.columns:
    print(" -", column)


# =====================================================================
# 5. IDENTIFY THE 42 BIOMARKERS
# =====================================================================

METADATA_COLUMNS = [
    "Patient ID #", "Patient ID", "Sample Type", "Cohort",
    "Sex", "Age", "AJCC Stage", "A/B/C", "T", "N", "M",
    "Histopathology"
]

if "Cohort" not in df.columns:
    raise ValueError("The dataset does not contain a 'Cohort' column.")

available_metadata = [
    col for col in METADATA_COLUMNS if col in df.columns
]

candidate_columns = [
    col for col in df.columns if col not in available_metadata
]

numeric_data = df[candidate_columns].apply(
    pd.to_numeric, errors="coerce"
)

biomarker_columns = [
    col for col in numeric_data.columns
    if numeric_data[col].notna().any()
]

print("\nNumber of identified biomarkers:", len(biomarker_columns))

if len(biomarker_columns) != EXPECTED_BIOMARKERS:
    raise ValueError(
        f"Expected {EXPECTED_BIOMARKERS} biomarkers, but identified "
        f"{len(biomarker_columns)}. Check the metadata exclusion list "
        "and the original Excel columns."
    )

print("\nBiomarkers included:")
for i, biomarker in enumerate(biomarker_columns, start=1):
    print(f"{i:02d}. {biomarker}")

df["Cohort"] = df["Cohort"].astype(str).str.strip()

print("\\nCOHORT-WISE DATA DETAILS")
print("-" * 80)
print(df["Cohort"].value_counts().reindex(COHORTS, fill_value=0).to_string())

print("\\nBIOMARKER DATA COMPLETENESS")
print("-" * 80)
for cohort in COHORTS:
    cohort_values = df.loc[
        df["Cohort"].str.lower() == cohort.lower(), biomarker_columns
    ].apply(pd.to_numeric, errors="coerce")
    non_missing = int(cohort_values.notna().sum().sum())
    total_values = int(cohort_values.shape[0] * cohort_values.shape[1])
    print(
        f"{cohort}: patients={len(cohort_values)}, "
        f"biomarkers={len(biomarker_columns)}, "
        f"non-missing={non_missing}/{total_values}, "
        f"missing={total_values - non_missing}"
    )

available_cohorts = {
    str(value).strip().lower()
    for value in df["Cohort"].dropna().unique()
}

for cohort in COHORTS:
    if cohort.lower() not in available_cohorts:
        raise ValueError(f"Cohort '{cohort}' was not found in the dataset.")


# =====================================================================
# 6. STORAGE DICTIONARIES
# =====================================================================

correlation_matrices = {}
dissimilarity_matrices = {}
persistence_results = {}
betti_results = {}
landscape_results = {}
cohort_sample_sizes = {}
persistence_interval_tables = {}


# =====================================================================
# 7. FUNCTION: PLOT PERSISTENCE DIAGRAM
# =====================================================================

def plot_persistence_diagram(persistence, cohort_name, save_dir):
    """
    H0: connected components.
    H1: one-dimensional loops.
    Infinite intervals are counted and annotated but not plotted
    as finite points because their death coordinate is infinite.
    """

    fig, ax = plt.subplots(figsize=(9.5, 8.8))

    h0_birth, h0_death = [], []
    h1_birth, h1_death = [], []
    essential_h0, essential_h1 = 0, 0

    for dimension, (birth, death) in persistence:
        if dimension not in (0, 1):
            continue

        if not np.isfinite(death):
            if dimension == 0:
                essential_h0 += 1
            elif dimension == 1:
                essential_h1 += 1
            continue

        if dimension == 0:
            h0_birth.append(birth)
            h0_death.append(death)
        elif dimension == 1:
            h1_birth.append(birth)
            h1_death.append(death)

    # Larger H0 points: hollow circles.
    if h0_birth:
        ax.scatter(
            h0_birth, h0_death,
            s=115,
            marker="o",
            facecolors="none",
            edgecolors="black",
            linewidths=2.0,
            label=r"$H_0$ (connected components)",
            zorder=3
        )

    # Larger H1 points: filled triangles.
    if h1_birth:
        ax.scatter(
            h1_birth, h1_death,
            s=125,
            marker="^",
            color="red",
            edgecolors="black",
            linewidths=1.0,
            label=r"$H_1$ (one-dimensional loops)",
            zorder=4
        )

    plot_max = MAX_EDGE_LENGTH

    ax.plot(
        [0, plot_max], [0, plot_max],
        linestyle="--",
        color="dimgray",
        linewidth=2.0,
        label="Diagonal",
        zorder=1
    )

    ax.set_xlim(0, plot_max)
    ax.set_ylim(0, plot_max)

    ax.set_xlabel("Birth", fontsize=AXIS_LABEL_SIZE, labelpad=12)
    ax.set_ylabel("Death", fontsize=AXIS_LABEL_SIZE, labelpad=12)
    ax.set_title(
        f"Persistence Diagram: {cohort_name}",
        fontsize=TITLE_SIZE,
        fontweight="bold",
        pad=18
    )

    annotation = (
        f"Essential $H_0$: {essential_h0}\n"
        f"Essential $H_1$: {essential_h1}"
    )

    ax.text(
        0.04, 0.96, annotation,
        transform=ax.transAxes,
        verticalalignment="top",
        horizontalalignment="left",
        fontsize=ANNOTATION_SIZE,
        bbox=dict(
            facecolor="white",
            edgecolor="black",
            linewidth=1.3,
            alpha=0.9,
            boxstyle="round,pad=0.45"
        )
    )

    ax.tick_params(
        axis="both", which="major",
        labelsize=TICK_LABEL_SIZE,
        width=1.6, length=7, direction="out"
    )

    ax.legend(
        loc="lower right",
        frameon=True,
        edgecolor="black",
        fontsize=LEGEND_SIZE,
        markerscale=1.15,
        borderpad=0.8,
        labelspacing=0.7
    )

    ax.grid(False)

    for spine in ax.spines.values():
        spine.set_linewidth(1.7)
        spine.set_color("black")

    plt.tight_layout()

    output_file = os.path.join(
        save_dir, f"Persistence_Diagram_{cohort_name}.png"
    )

    plt.savefig(
        output_file,
        dpi=600,
        bbox_inches="tight",
        facecolor="white"
    )

    print("Persistence diagram saved:", output_file)
    plt.show()
    plt.close(fig)


# =====================================================================
# 8. FUNCTION: CALCULATE PERSISTENCE LANDSCAPES
# =====================================================================

def calculate_persistence_landscape(intervals, radii, n_layers=5):
    """
    For each finite interval [birth, death], calculate
    f(t) = max(0, min(t - birth, death - t)).
    The kth landscape is the kth largest tent value at each t.
    Infinite intervals are excluded because they do not define
    finite tent functions.
    """

    finite_intervals = [
        (birth, death)
        for birth, death in intervals
        if (
            np.isfinite(birth)
            and np.isfinite(death)
            and death > birth
        )
    ]

    if not finite_intervals:
        return np.zeros((n_layers, len(radii)), dtype=float)

    intervals_array = np.asarray(finite_intervals, dtype=float)
    births = intervals_array[:, 0]
    deaths = intervals_array[:, 1]

    tent_values = np.minimum(
        radii[np.newaxis, :] - births[:, np.newaxis],
        deaths[:, np.newaxis] - radii[np.newaxis, :]
    )

    tent_values = np.maximum(tent_values, 0.0)
    sorted_values = np.sort(tent_values, axis=0)[::-1]

    landscapes = np.zeros((n_layers, len(radii)), dtype=float)
    available_layers = min(n_layers, sorted_values.shape[0])

    landscapes[:available_layers, :] = (
        sorted_values[:available_layers, :]
    )

    return landscapes


# =====================================================================
# 9. FUNCTION: PLOT AND SAVE H0/H1 LANDSCAPES SEPARATELY
# =====================================================================

def plot_single_persistence_landscape(
    landscape,
    cohort_name,
    dimension_label,
    radii,
    save_dir
):
    """Save one homological dimension per figure."""

    fig, ax = plt.subplots(figsize=(10.5, 7.8))

    layer_styles = ["-", "--", "-.", ":", (0, (5, 1))]
    layer_widths = [3.4, 3.0, 2.8, 2.8, 2.8]

    for k in range(landscape.shape[0]):
        ax.plot(
            radii,
            landscape[k, :],
            linestyle=layer_styles[k % len(layer_styles)],
            linewidth=layer_widths[k % len(layer_widths)],
            label=rf"$\lambda_{k+1}(t)$"
        )

    ax.set_xlabel(
        "Filtration Parameter",
        fontsize=AXIS_LABEL_SIZE,
        labelpad=12
    )
    ax.set_ylabel(
        "Landscape Value",
        fontsize=AXIS_LABEL_SIZE,
        labelpad=12
    )
    ax.set_title(
        rf"{dimension_label} Persistence Landscape: {cohort_name}",
        fontsize=TITLE_SIZE,
        fontweight="bold",
        pad=18
    )

    ax.set_xlim(0, MAX_EDGE_LENGTH)

    ax.tick_params(
        axis="both", which="major",
        labelsize=TICK_LABEL_SIZE,
        width=1.6, length=7, direction="out"
    )

    ax.legend(
        loc="best",
        frameon=True,
        edgecolor="black",
        fontsize=LEGEND_SIZE,
        borderpad=0.8,
        labelspacing=0.7
    )

    ax.grid(False)

    for spine in ax.spines.values():
        spine.set_linewidth(1.7)
        spine.set_color("black")

    plt.tight_layout()

    output_file = os.path.join(
        save_dir,
        f"Persistence_Landscape_{dimension_label}_{cohort_name}.png"
    )

    plt.savefig(
        output_file,
        dpi=600,
        bbox_inches="tight",
        facecolor="white"
    )

    print(f"{dimension_label} landscape saved:", output_file)
    plt.show()
    plt.close(fig)


# =====================================================================
# 10. FUNCTION: CALCULATE BETTI CURVES
# =====================================================================

def calculate_betti_curves(persistence, radii):
    """
    A feature is alive at r if birth <= r < death.
    Infinite intervals are included.
    """

    betti_0 = np.zeros(len(radii), dtype=int)
    betti_1 = np.zeros(len(radii), dtype=int)

    for dimension, (birth, death) in persistence:
        if dimension not in (0, 1):
            continue

        target = betti_0 if dimension == 0 else betti_1
        alive = (radii >= birth) & (radii < death)
        target[alive] += 1

    return betti_0, betti_1


# =====================================================================
# PART 1: COMPUTE PERSISTENCE AND SAVE DIAGRAMS
# =====================================================================

print("\n" + "=" * 80)
print("PART 1: PERSISTENCE DIAGRAM CONSTRUCTION")
print("=" * 80)

for cohort in COHORTS:

    print("\n" + "-" * 80)
    print(f"PROCESSING COHORT: {cohort}")
    print("-" * 80)

    # 11.1 Extract cohort-specific patient data.
    cohort_df = df.loc[
        df["Cohort"].str.lower() == cohort.lower(),
        biomarker_columns
    ].copy()

    cohort_df = cohort_df.apply(pd.to_numeric, errors="coerce")
    cohort_df = cohort_df.replace([np.inf, -np.inf], np.nan)
    cohort_df = cohort_df.dropna(how="all")

    if len(cohort_df) < 2:
        raise ValueError(f"Insufficient patients in {cohort} cohort.")

    cohort_sample_sizes[cohort] = len(cohort_df)
    print("Number of patients:", len(cohort_df))

    # 11.2 Calculate correlation matrix.
    correlation_matrix = cohort_df.corr(
        method="pearson",
        min_periods=2
    )

    correlation_matrix = correlation_matrix.fillna(0)
    correlation_matrix = (
        correlation_matrix + correlation_matrix.T
    ) / 2
    correlation_matrix = correlation_matrix.clip(-1, 1)

    np.fill_diagonal(correlation_matrix.values, 1.0)
    correlation_matrices[cohort] = correlation_matrix

    print("Correlation matrix shape:", correlation_matrix.shape)

    # 11.3 Calculate dissimilarity matrix.
    dissimilarity_matrix = 1.0 - np.abs(correlation_matrix.values)
    dissimilarity_matrix = np.clip(dissimilarity_matrix, 0.0, 1.0)
    dissimilarity_matrix = (
        dissimilarity_matrix + dissimilarity_matrix.T
    ) / 2

    np.fill_diagonal(dissimilarity_matrix, 0.0)
    dissimilarity_matrices[cohort] = dissimilarity_matrix

    print("Dissimilarity matrix shape:", dissimilarity_matrix.shape)

    # 11.4 Construct Vietoris-Rips complex.
    print("Constructing Vietoris-Rips complex...")

    rips_complex = gd.RipsComplex(
        distance_matrix=dissimilarity_matrix,
        max_edge_length=MAX_EDGE_LENGTH
    )

    simplex_tree = rips_complex.create_simplex_tree(
        max_dimension=MAX_HOMOLOGY_DIMENSION
    )

    print("Number of simplices:", simplex_tree.num_simplices())

    # 11.5 Calculate persistent homology.
    print("Calculating persistent homology...")

    persistence = simplex_tree.persistence(
        homology_coeff_field=HOMOLOGY_COEFFICIENT_FIELD,
        min_persistence=MIN_PERSISTENCE
    )

    persistence_results[cohort] = persistence

    # 11.6 Extract and print persistence intervals.
    h0_intervals = [
        interval for dimension, interval in persistence
        if dimension == 0
    ]

    h1_intervals = [
        interval for dimension, interval in persistence
        if dimension == 1
    ]

    print("\\nPERSISTENCE INTERVAL DETAILS")
    print("H0 intervals:", len(h0_intervals))
    print("H1 intervals:", len(h1_intervals))
    print(
        "Finite H0 intervals:",
        sum(1 for birth, death in h0_intervals if np.isfinite(death))
    )
    print(
        "Finite H1 intervals:",
        sum(1 for birth, death in h1_intervals if np.isfinite(death))
    )

    print("\nTotal H0 intervals:", len(h0_intervals))
    print("Total H1 intervals:", len(h1_intervals))

    interval_rows = []

    for dimension, (birth, death) in persistence:
        interval_rows.append({
            "Cohort": cohort,
            "Dimension": dimension,
            "Birth": birth,
            "Death": death,
            "Persistence": (
                death - birth if np.isfinite(death) else np.inf
            )
        })

    persistence_interval_tables[cohort] = pd.DataFrame(interval_rows)

    # 11.7 Generate and save persistence diagram.
    print("\nGenerating persistence diagram...")

    plot_persistence_diagram(
        persistence=persistence,
        cohort_name=cohort,
        save_dir=OUTPUT_DIR
    )


print("\n" + "=" * 80)
print("ALL FOUR PERSISTENCE DIAGRAMS HAVE BEEN GENERATED.")
print("=" * 80)


# =====================================================================
# PART 2: CONSTRUCT, PRINT AND SAVE H0/H1 LANDSCAPES SEPARATELY
# =====================================================================

print("\n" + "=" * 80)
print("PART 2: PERSISTENCE LANDSCAPE CONSTRUCTION")
print("=" * 80)

for cohort in COHORTS:

    print("\n" + "-" * 80)
    print(f"CONSTRUCTING LANDSCAPES: {cohort}")
    print("-" * 80)

    persistence = persistence_results[cohort]

    h0_intervals = [
        interval for dimension, interval in persistence
        if dimension == 0
    ]

    h1_intervals = [
        interval for dimension, interval in persistence
        if dimension == 1
    ]

    # 12.1 Calculate H0 and H1 landscapes.
    h0_landscape = calculate_persistence_landscape(
        intervals=h0_intervals,
        radii=RADII,
        n_layers=N_LANDSCAPE_LAYERS
    )

    h1_landscape = calculate_persistence_landscape(
        intervals=h1_intervals,
        radii=RADII,
        n_layers=N_LANDSCAPE_LAYERS
    )

    landscape_results[cohort] = {
        "H0": h0_landscape,
        "H1": h1_landscape
    }

    # 12.2 Print numerical landscape information.
    print("\nH0 PERSISTENCE LANDSCAPE")
    print("-" * 55)

    for k in range(N_LANDSCAPE_LAYERS):
        print(
            f"Lambda {k + 1}: Maximum value = "
            f"{np.max(h0_landscape[k, :]):.6f}"
        )

    print("\nH1 PERSISTENCE LANDSCAPE")
    print("-" * 55)

    for k in range(N_LANDSCAPE_LAYERS):
        print(
            f"Lambda {k + 1}: Maximum value = "
            f"{np.max(h1_landscape[k, :]):.6f}"
        )

    # 12.3 Save H0 landscape separately.
    plot_single_persistence_landscape(
        landscape=h0_landscape,
        cohort_name=cohort,
        dimension_label="H0",
        radii=RADII,
        save_dir=OUTPUT_DIR
    )

    # 12.4 Save H1 landscape separately.
    plot_single_persistence_landscape(
        landscape=h1_landscape,
        cohort_name=cohort,
        dimension_label="H1",
        radii=RADII,
        save_dir=OUTPUT_DIR
    )


print("\n" + "=" * 80)
print("ALL H0 AND H1 PERSISTENCE LANDSCAPES HAVE BEEN GENERATED.")
print("=" * 80)


# =====================================================================
# PART 3: CONSTRUCT AND SAVE COMBINED BETTI CURVES
# =====================================================================

print("\n" + "=" * 80)
print("PART 3: BETTI CURVE CONSTRUCTION")
print("=" * 80)

# 13.1 Calculate Betti curves for all cohorts.
for cohort in COHORTS:

    betti_0, betti_1 = calculate_betti_curves(
        persistence=persistence_results[cohort],
        radii=RADII
    )

    betti_results[cohort] = {
        "Betti_0": betti_0,
        "Betti_1": betti_1
    }

    print(f"\n{cohort} cohort:")
    print("Maximum Betti-0:", np.max(betti_0))
    print("Maximum Betti-1:", np.max(betti_1))
    print("Final Betti-0:", betti_0[-1])
    print("Final Betti-1:", betti_1[-1])


# 13.2 Cohort plotting styles.
COHORT_STYLES = {
    "Normal": {
        "color": "black",
        "linestyle": "-",
        "linewidth": 3.2
    },
    "Bladder": {
        "color": "red",
        "linestyle": "--",
        "linewidth": 3.2
    },
    "Ovarian": {
        "color": "blue",
        "linestyle": "-.",
        "linewidth": 3.2
    },
    "Pancreatic": {
        "color": "green",
        "linestyle": ":",
        "linewidth": 3.5
    }
}


# 13.3 Plot combined Betti curves.
fig, axes = plt.subplots(
    nrows=1,
    ncols=2,
    figsize=(18, 8)
)

# Betti-0.
ax = axes[0]

for cohort in COHORTS:
    style = COHORT_STYLES[cohort]

    ax.plot(
        RADII,
        betti_results[cohort]["Betti_0"],
        color=style["color"],
        linestyle=style["linestyle"],
        linewidth=style["linewidth"],
        label=cohort
    )

ax.set_xlabel(
    "Filtration Radius",
    fontsize=AXIS_LABEL_SIZE,
    labelpad=12
)
ax.set_ylabel(
    r"Betti Number $\beta_0$",
    fontsize=AXIS_LABEL_SIZE,
    labelpad=12
)
ax.set_title(
    r"Betti-0 Curves ($H_0$)",
    fontsize=TITLE_SIZE,
    fontweight="bold",
    pad=18
)
ax.set_xlim(0, MAX_EDGE_LENGTH)
ax.tick_params(
    axis="both", which="major",
    labelsize=TICK_LABEL_SIZE,
    width=1.6, length=7, direction="out"
)
ax.legend(
    loc="best",
    frameon=True,
    edgecolor="black",
    fontsize=LEGEND_SIZE,
    borderpad=0.8
)
ax.grid(False)


# Betti-1.
ax = axes[1]

for cohort in COHORTS:
    style = COHORT_STYLES[cohort]

    ax.plot(
        RADII,
        betti_results[cohort]["Betti_1"],
        color=style["color"],
        linestyle=style["linestyle"],
        linewidth=style["linewidth"],
        label=cohort
    )

ax.set_xlabel(
    "Filtration Radius",
    fontsize=AXIS_LABEL_SIZE,
    labelpad=12
)
ax.set_ylabel(
    r"Betti Number $\beta_1$",
    fontsize=AXIS_LABEL_SIZE,
    labelpad=12
)
ax.set_title(
    r"Betti-1 Curves ($H_1$)",
    fontsize=TITLE_SIZE,
    fontweight="bold",
    pad=18
)
ax.set_xlim(0, MAX_EDGE_LENGTH)
ax.tick_params(
    axis="both", which="major",
    labelsize=TICK_LABEL_SIZE,
    width=1.6, length=7, direction="out"
)
ax.legend(
    loc="best",
    frameon=True,
    edgecolor="black",
    fontsize=LEGEND_SIZE,
    borderpad=0.8
)
ax.grid(False)

for ax in axes:
    for spine in ax.spines.values():
        spine.set_linewidth(1.7)
        spine.set_color("black")

plt.tight_layout(pad=2.5)

combined_betti_file = os.path.join(
    OUTPUT_DIR,
    "Combined_Betti_Curves_All_Cohorts.png"
)

plt.savefig(
    combined_betti_file,
    dpi=600,
    bbox_inches="tight",
    facecolor="white"
)

print("\nCombined Betti curves saved:", combined_betti_file)

plt.show()
plt.close(fig)


# =====================================================================
# PART 4: PRINT AND SAVE FINAL PERSISTENT HOMOLOGY SUMMARY
# =====================================================================

print("\n" + "=" * 80)
print("PART 4: FINAL PERSISTENT HOMOLOGY SUMMARY")
print("=" * 80)

summary_rows = []

for cohort in COHORTS:

    persistence = persistence_results[cohort]

    h0_intervals = [
        interval for dimension, interval in persistence
        if dimension == 0
    ]

    h1_intervals = [
        interval for dimension, interval in persistence
        if dimension == 1
    ]

    betti_0 = betti_results[cohort]["Betti_0"]
    betti_1 = betti_results[cohort]["Betti_1"]

    h0_landscape = landscape_results[cohort]["H0"]
    h1_landscape = landscape_results[cohort]["H1"]

    summary_rows.append({
        "Cohort": cohort,
        "Number of Patients": cohort_sample_sizes[cohort],
        "Number of Biomarkers": EXPECTED_BIOMARKERS,
        "Number of H0 Intervals": len(h0_intervals),
        "Number of H1 Intervals": len(h1_intervals),
        "Essential H0": sum(
            1 for birth, death in h0_intervals
            if not np.isfinite(death)
        ),
        "Essential H1": sum(
            1 for birth, death in h1_intervals
            if not np.isfinite(death)
        ),
        "Maximum Betti-0": int(np.max(betti_0)),
        "Maximum Betti-1": int(np.max(betti_1)),
        "Maximum H0 Landscape Lambda1": float(
            np.max(h0_landscape[0, :])
        ),
        "Maximum H1 Landscape Lambda1": float(
            np.max(h1_landscape[0, :])
        )
    })

summary_df = pd.DataFrame(summary_rows)
print(summary_df.to_string(index=False))

summary_file = os.path.join(
    OUTPUT_DIR,
    "Persistent_Homology_Summary_All_Cohorts.xlsx"
)

summary_df.to_excel(summary_file, index=False)
print("\nSummary file saved:", summary_file)


# =====================================================================
# PART 5: SAVE NUMERICAL BETTI CURVE VALUES
# =====================================================================

betti_data = {"Filtration Radius": RADII}

for cohort in COHORTS:
    betti_data[f"{cohort}_Betti_0"] = (
        betti_results[cohort]["Betti_0"]
    )
    betti_data[f"{cohort}_Betti_1"] = (
        betti_results[cohort]["Betti_1"]
    )

betti_df = pd.DataFrame(betti_data)

betti_file = os.path.join(
    OUTPUT_DIR,
    "Betti_Curve_Values_All_Cohorts.xlsx"
)

betti_df.to_excel(betti_file, index=False)
print("\nBetti curve values saved:", betti_file)


# =====================================================================
# PART 6: SAVE NUMERICAL PERSISTENCE LANDSCAPE VALUES
# =====================================================================

landscape_file = os.path.join(
    OUTPUT_DIR,
    "Persistence_Landscape_Values_All_Cohorts.xlsx"
)

with pd.ExcelWriter(landscape_file, engine="openpyxl") as writer:

    for cohort in COHORTS:

        h0_landscape = landscape_results[cohort]["H0"]
        h1_landscape = landscape_results[cohort]["H1"]

        landscape_data = {
            "Filtration Parameter": RADII
        }

        for k in range(N_LANDSCAPE_LAYERS):
            landscape_data[f"H0_Lambda_{k + 1}"] = (
                h0_landscape[k, :]
            )
            landscape_data[f"H1_Lambda_{k + 1}"] = (
                h1_landscape[k, :]
            )

        landscape_df = pd.DataFrame(landscape_data)

        landscape_df.to_excel(
            writer,
            sheet_name=cohort,
            index=False
        )

print("\nPersistence landscape values saved:", landscape_file)


# =====================================================================
# PART 7: SAVE PERSISTENCE INTERVALS
# =====================================================================

persistence_intervals_file = os.path.join(
    OUTPUT_DIR,
    "Persistence_Intervals_All_Cohorts.xlsx"
)

with pd.ExcelWriter(
    persistence_intervals_file,
    engine="openpyxl"
) as writer:

    for cohort in COHORTS:
        persistence_interval_tables[cohort].to_excel(
            writer,
            sheet_name=cohort,
            index=False
        )

print("\nPersistence intervals saved:", persistence_intervals_file)


# =====================================================================
# PART 8: SAVE CORRELATION AND DISSIMILARITY MATRICES
# =====================================================================

matrices_file = os.path.join(
    OUTPUT_DIR,
    "Correlation_and_Dissimilarity_Matrices.xlsx"
)

with pd.ExcelWriter(matrices_file, engine="openpyxl") as writer:

    for cohort in COHORTS:

        correlation_matrices[cohort].to_excel(
            writer,
            sheet_name=f"{cohort}_Correlation"
        )

        dissimilarity_df = pd.DataFrame(
            dissimilarity_matrices[cohort],
            index=biomarker_columns,
            columns=biomarker_columns
        )

        dissimilarity_df.to_excel(
            writer,
            sheet_name=f"{cohort}_Dissimilarity"
        )

print("\nCorrelation and dissimilarity matrices saved:", matrices_file)


# =====================================================================
# PART 9: FINAL OUTPUT
# =====================================================================

print("\n" + "=" * 80)
print("ANALYSIS COMPLETED SUCCESSFULLY")
print("=" * 80)

print("\nAll files are saved in:")
print(OUTPUT_DIR)

print("\nGENERATED PERSISTENCE DIAGRAMS:")
for cohort in COHORTS:
    print(f"Persistence_Diagram_{cohort}.png")

print("\nGENERATED PERSISTENCE LANDSCAPES (SAVED SEPARATELY):")
for cohort in COHORTS:
    print(f"Persistence_Landscape_H0_{cohort}.png")
    print(f"Persistence_Landscape_H1_{cohort}.png")

print("\nOTHER FIGURES:")
print("Combined_Betti_Curves_All_Cohorts.png")

print("\nGENERATED EXCEL FILES:")
print("Persistent_Homology_Summary_All_Cohorts.xlsx")
print("Betti_Curve_Values_All_Cohorts.xlsx")
print("Persistence_Landscape_Values_All_Cohorts.xlsx")
print("Persistence_Intervals_All_Cohorts.xlsx")
print("Correlation_and_Dissimilarity_Matrices.xlsx")

print("\nAll four cohorts were analysed using all 42 biomarkers.")
print("=" * 80)
```
